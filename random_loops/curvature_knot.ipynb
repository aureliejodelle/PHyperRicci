```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

from curvature_viz import load_real_data, run_all

pio.renderers.default = "browser"
```

```python
Df = load_real_data(outputs_dir="outputs")
Df
```

```python
fig = px.box(Df, y="Curvature", color="Type", x = 'Length', notched=True, boxmode='overlay',
                   )
#fig.show()
```

```python
for L in [100,150,200,250,300,350,400,450,500]:


    fig = px.histogram(Df.loc[Df["Length"] ==L], x="Curvature", color="Type", 
                       marginal="box", # or violin, rug
                       barmode='overlay',
                       histnorm='probability',
                       hover_data=Df.columns)
    #fig.show()
```

```python
run_all(Df, out_dir="my_plots")
```

```python
L = 500

for j in range(10):
    unknots = {}
    knots = {}

    while len(unknots) < 50 or len(knots) < 50:
        K = generate_loop(L, 1, output='list') 
        poly = jones(K[0], closure=0) 
        if poly == '0_1'  and len(unknots) <50:
            i = len(unknots)
            unknots[i] = [K[0], poly]
        if poly != '0_1' and len(knots) < 50:
            i = len(knots)
            knots[i] = [K[0], poly] 
            #print(poly)
    with open("knots_jodelle_{}_{}.json".format(L,j), 'w') as f:
        json.dump(knots,f)
    with open("unknots_jodelle_{}_{}.json".format(L,j), 'w') as f:
        json.dump(unknots,f)
```
